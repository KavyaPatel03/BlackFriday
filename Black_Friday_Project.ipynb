```python
#Question 1 - How many male and female shoppers are there?
data['Gender'].value_counts()
```

```python
g1 = data['Gender'].value_counts()
g1.plot(kind = 'bar')
plt.ylabel('Count')
plt.show()
```

Observations :-
Total number of male shoppers are 4,14,259.
Total number of female shoppers are 1,35,809.

```python
#Question 2 - What is the most common age group of shoppers?
data['Age'].value_counts()
```

```python
g1 = data['Age'].value_counts()
expl = [0.2,0,0,0,0,0,0]
g1.plot(kind = 'pie',autopct = '%6.2F%%',explode = expl)
plt.show()
```

Observations :- The most common age group of shoppers is 26-35.

```python
#Question 3 - How many shoppers are from different city categories?
data['City_Category'].value_counts()
```

```python
g1 = data['City_Category'].value_counts()
color = ['r','y','b']
g1.plot(kind = 'barh', color =color)
plt.show()
```

Observations :- The most number of shoppers are from city B.

```python
#Question 4 - How many shoppers are married?
data[data["Marital_Status"] == "married"]["User_ID"].nunique()
```

Observation :- There are 2474 unique married shoppers.

```python
#Question 5 - How many shoppers are unmarried?
data[data["Marital_Status"] == "unmarried"]["User_ID"].nunique()
```

Observation :- There are 3417 unique unmarried shoppers.

```python
married_count = data[data["Marital_Status"] == "married"]["User_ID"].nunique()
unmarried_count = data[data["Marital_Status"] == "unmarried"]["User_ID"].nunique()

marital_status_counts = pd.Series({'Married': married_count, 'Unmarried': unmarried_count})
marital_status_counts.plot(kind = 'bar',color = 'b')
plt.ylabel('Number of Users')
plt.xlabel('Marital Status')
plt.show()
```

```python
#Question 6 - What is the total amount of purchase by each gender?
data[data['Gender'] == 'Male']['Purchase'].sum()
```

```python
data[data['Gender'] == 'Female']['Purchase'].sum()
```

```python
g1 = data[data['Gender'] == 'Male']['Purchase'].sum()
g2 = data[data['Gender'] == 'Female']['Purchase'].sum()
g3 = pd.Series({'Male': g1, 'Female': g2})
g4 = ['r','b']
g3.plot(kind = 'bar',color = g4)
plt.ylabel('Purchase')
plt.xlabel('Gender')
plt.show()
```

Observation :-
The total purchase by Male is 1,81,12,44,855.
The total purchase by Female is 54,89,30,546.

```python
#Question 7 - What is the total amount of purchase by each age category?
data.groupby('Age')['Purchase'].sum().sort_values(ascending = False)
```

```python
g1 = data.groupby('Age')['Purchase'].sum().sort_values(ascending = False)
g1.plot(kind = 'pie',autopct = '%6.2f%%')
plt.show()
```

Observation :- The above result shows the total amount of purchase by each age category.

```python
#Question 8 -  What is the total amount of purchase done by each category of 'Stay_In_Current_City_Years'?
data.groupby('Stay_In_Current_City_Years')['Purchase'].sum().sort_values(ascending = False)
```

```python
g1 = data.groupby('Stay_In_Current_City_Years')['Purchase'].sum().sort_values(ascending = False)
g1.plot(kind = 'pie',autopct = '%6.2f%%')
plt.show()
```

Observation :- The above result shows that how much purchase has been done by the people who are living in that area.

```python
#Question 9 - Which product category has sold the highest amount of goods?
data['Product_Category_1'].sum()
```

```python
data['Product_Category_2'].sum()
```

```python
data['Product_Category_3'].sum()
```

```python
g1 = data['Product_Category_1'].sum()
g2 = data['Product_Category_2'].sum()
g3 = data['Product_Category_3'].sum()
g4 = pd.Series({'Product_Category_1': g1, 'Product_Category_2': g2, 'Product_Category_3': g3})
g4.plot(kind = 'bar')
plt.show()
```

```python
#Question 10 - What is the highest frequency of the a product being purchased?
data['Product_ID'].value_counts().sort_values(ascending=False)
```

```python
g1 = data['Product_ID'].value_counts().reset_index().head(5)
g1.plot(kind = 'scatter',x = 'Product_ID',y = 'count')
plt.show()
```

Observation :- The product which has product id 'P00025442' has been purchased the most number of times.

```python
#Question 11 - What are the top 10 highest spending unique 'User_id'?
data.groupby('User_ID')['Purchase'].sum().sort_values(ascending=False).reset_index().head(10)
```

```python
g1 = data.groupby('User_ID')['Purchase'].sum().sort_values(ascending=False).reset_index().head(10)
g1.plot(kind = 'scatter',x = 'User_ID',y = 'Purchase')
plt.show()
```

Observation :- The above result shows the top 10 most spending user_id.

```python
#Question 12 - What are the least 10 highest spending unique 'User_id'?
data.groupby('User_ID')['Purchase'].sum().sort_values(ascending=False).reset_index().tail(10)
```

```python
g1 = data.groupby('User_ID')['Purchase'].sum().sort_values(ascending=False).reset_index().tail(10)
g1.plot(kind = 'scatter',x = 'User_ID',y = 'Purchase')
plt.show()
```

Observation :- The above result shows the top 10 least spending user_id.
